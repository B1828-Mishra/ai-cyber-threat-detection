```python
import pandas as pd

base_path = r"C:\Users\KIIT0001\Desktop\ai-cyber-threat-detection\dataset\MachineLearningCSV\MachineLearningCVE"

files = [
    "Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday-WorkingHours-Morning.pcap_ISCX.csv"
]

data = []

for file in files:
    path = base_path + "\\" + file
    
    temp = pd.read_csv(path)
    temp.columns = temp.columns.str.strip()
    
    # Take a maximum of 20,000 rows from each file
    temp = temp.sample(
        n=min(20000, len(temp)),
        random_state=42
    )
    
    data.append(temp)

df = pd.concat(data, ignore_index=True)

print("Combined shape:", df.shape)
print("\nLabels:")
print(df["Label"].value_counts())
```

```python
df["Target"] = df["Label"].apply(
    lambda x: 0 if x == "BENIGN" else 1
)

print(df[["Label", "Target"]].head())

print("\nTarget distribution:")
print(df["Target"].value_counts())
```

```python
# Remove the original text label
df = df.drop(columns=["Label"])

# Replace infinite values with NaN
df = df.replace([float("inf"), float("-inf")], pd.NA)

# Check missing values
print("Missing values before cleaning:")
print(df.isna().sum().sum())

# Remove rows containing missing values
df = df.dropna()

print("\nShape after cleaning:", df.shape)

print("\nTarget distribution after cleaning:")
print(df["Target"].value_counts())
```

```python
# Features
X = df.drop(columns=["Target"])

# Target
y = df["Target"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature data types:")
print(X.dtypes.value_counts())
```

```python
# Find non-numeric columns
non_numeric = X.select_dtypes(include=["object"]).columns

print("Non-numeric columns:")
print(list(non_numeric))

print("\nSample values:")
for col in non_numeric:
    print(f"\n{col}:")
    print(X[col].head())
```

```python
for col in ["Flow Bytes/s", "Flow Packets/s"]:
    X[col] = pd.to_numeric(X[col], errors="coerce")

print("Feature data types after conversion:")
print(X.dtypes.value_counts())

print("\nMissing values after conversion:")
print(X.isna().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Training model...")

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
# Make predictions on unseen test data
y_pred = model.predict(X_test)

print("Predictions completed!")

print("\nFirst 20 predictions:")
print(y_pred[:20])

print("\nActual values:")
print(y_test.values[:20])
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Model Performance")
print("------------------")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["BENIGN", "ATTACK"]
)

disp.plot()
plt.title("Cyber Threat Detection - Confusion Matrix")
plt.show()
```

```python
import os
import joblib
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split


# -----------------------------
# 1. Paths
# -----------------------------

BASE_PATH = r"C:\Users\KIIT0001\Desktop\ai-cyber-threat-detection"

DATA_PATH = os.path.join(
    BASE_PATH,
    "dataset",
    "MachineLearningCSV",
    "MachineLearningCVE"
)

MODEL_PATH = os.path.join(
    BASE_PATH,
    "models",
    "random_forest_model.pkl"
)

FEATURE_PATH = os.path.join(
    BASE_PATH,
    "models",
    "feature_names.pkl"
)


# -----------------------------
# 2. Dataset files
# -----------------------------

files = [
    "Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday-WorkingHours-Morning.pcap_ISCX.csv"
]


# -----------------------------
# 3. Load dataset
# -----------------------------

data = []

for file in files:

    path = os.path.join(DATA_PATH, file)

    temp = pd.read_csv(path)

    temp.columns = temp.columns.str.strip()

    # Take a manageable sample
    temp = temp.sample(
        n=min(20000, len(temp)),
        random_state=42
    )

    data.append(temp)


df = pd.concat(data, ignore_index=True)

print("Dataset shape:", df.shape)


# -----------------------------
# 4. Create binary target
# -----------------------------

df["Target"] = df["Label"].apply(
    lambda x: 0 if x == "BENIGN" else 1
)

df = df.drop(columns=["Label"])

# Replace infinite values
df = df.replace(
    [float("inf"), float("-inf")],
    pd.NA
)

df = df.dropna()


# -----------------------------
# 5. Separate features
# -----------------------------

X = df.drop(columns=["Target"])
y = df["Target"]


# Convert flow rate columns
for col in ["Flow Bytes/s", "Flow Packets/s"]:

    X[col] = pd.to_numeric(
        X[col],
        errors="coerce"
    )

X = X.dropna()


# -----------------------------
# 6. Train-test split
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# -----------------------------
# 7. Train Random Forest
# -----------------------------

print("Training Random Forest...")

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("Training completed!")


# -----------------------------
# 8. Save model
# -----------------------------

os.makedirs(
    os.path.dirname(MODEL_PATH),
    exist_ok=True
)

joblib.dump(model, MODEL_PATH)

joblib.dump(
    list(X.columns),
    FEATURE_PATH
)

print("Model saved to:")
print(MODEL_PATH)

print("Feature names saved to:")
print(FEATURE_PATH)
```
